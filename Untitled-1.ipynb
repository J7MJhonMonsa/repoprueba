```python
import numpy as np
import matplotlib.pyplot as plt

print("intento 3")

# --- Configuración de la señal ---
fs = 1000       # Frecuencia de muestreo (Hz)
duracion = 1.0  # Duración en segundos
t = np.linspace(0, duracion, int(fs * duracion), endpoint=False)
n_muestras = len(t)

# 1. Ruido Blanco Gaussiano (media = 0, desviación estándar = 1)
ruido_blanco = np.random.normal(0, 1, n_muestras)

# 2. Ruido Rosa (1/f) - Generado mediante filtrado en el dominio de la frecuencia
frecuencias = np.fft.rfftfreq(n_muestras, d=1/fs)
# Evitar división por cero en DC (frecuencia 0)
frecuencias[0] = 1.0 
espectro_blanco = np.fft.rfft(ruido_blanco)
# Escalar el espectro por 1/sqrt(f) para obtener una densidad espectral de potencia 1/f
espectro_rosa = espectro_blanco / np.sqrt(frecuencias)
ruido_rosa = np.fft.irfft(espectro_rosa, n=n_muestras)
# Normalizar amplitud
ruido_rosa = ruido_rosa / np.max(np.abs(ruido_rosa))

# 3. Ruido de Impulso / Sal y Pimienta (picos aleatorios)
ruido_impulso = np.zeros(n_muestras)
probabilidad_impulso = 0.03  # 3% de muestras afectadas
mascara = np.random.rand(n_muestras) < probabilidad_impulso
ruido_impulso[mascara] = np.random.choice([-1.0, 1.0], size=np.sum(mascara))

# --- Visualización ---
plt.figure(figsize=(12, 8))

plt.subplot(3, 1, 1)
plt.plot(t, ruido_blanco, color='crimson', lw=0.8)
plt.title('Ruido Blanco Gaussiano (Distribución Normal)')
plt.ylabel('Amplitud')
plt.grid(True, alpha=0.3)

plt.subplot(3, 1, 2)
plt.plot(t, ruido_rosa, color='magenta', lw=0.8)
plt.title('Ruido Rosa (Densidad Espectral de Potencia 1/f)')
plt.ylabel('Amplitud')
plt.grid(True, alpha=0.3)

plt.subplot(3, 1, 3)
plt.plot(t, ruido_impulso, color='darkorange', lw=1.2)
plt.title('Ruido de Impulso (Picos Esporádicos)')
plt.xlabel('Tiempo [s]')
plt.ylabel('Amplitud')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
